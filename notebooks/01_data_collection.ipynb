```python
import nflreadpy as nfl
import pandas as pd
```

```python
seasons = list(range(2015,2027))

games = nfl.load_schedules(seasons).to_pandas()
weekly = nfl.load_player_stats(seasons).to_pandas()
```

```python
games.to_csv('../data/raw_games.csv', index=False)
weekly.to_csv('../data/raw_weekly.csv', index=False)
```
